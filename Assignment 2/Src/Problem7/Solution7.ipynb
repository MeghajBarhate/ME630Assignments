> ## **Convective quench of a short cylinder (2D axisymmetric r−z)**

# Problem 7: Convective Quench of a Short Cylinder

Consider transient heat conduction in a **short cylinder** with two-dimensional axisymmetric heat transfer in the \(r-z\) plane.

The cylinder has:

* Radius:

  $$
  R = 0.04\ \text{m}
  $$

* Half-length:

  $$
  L_z = 0.03\ \text{m}
  $$

* Thermal diffusivity:

  $$
  \alpha = 1.0\times10^{-5}\ \text{m}^2/\text{s}
  $$

* Thermal conductivity:

  $$
  k = 20\ \text{W/(m K)}
  $$

Initially, the entire cylinder is at a uniform temperature

$$
T_i = 250^\circ\text{C}.
$$

At \(t=0\), the cylinder is suddenly exposed to a surrounding fluid at

$$
T_\infty = 25^\circ\text{C},
$$

with a convection heat-transfer coefficient

$$
h = 150\ \text{W/(m}^2\text{K)}.
$$

Convection occurs from the **curved cylindrical surface** as well as from the **two end faces**.

The corresponding Biot numbers are

$$
Bi_R = \frac{hR}{k} = 0.30,
$$

and

$$
Bi_z = \frac{hL_z}{k} = 0.225.
$$

---

## Governing Equation

For axisymmetric transient conduction with no internal heat generation, the governing equation is

$$
\frac{\partial T}{\partial t}
=
\alpha
\left[
\frac{\partial^2 T}{\partial r^2}
+
\frac{1}{r}\frac{\partial T}{\partial r}
+
\frac{\partial^2 T}{\partial z^2}
\right].
$$

The solution is independent of the angular coordinate \(\phi\).

---

## Initial Condition

The cylinder is initially at a uniform temperature:

$$
T(r,z,0)=T_i.
$$

---

## Boundary Conditions

### Axis of symmetry

At the cylinder axis,

$$
\left.
\frac{\partial T}{\partial r}
\right|_{r=0}=0.
$$

### Curved surface

At \(r=R\), convection occurs between the cylinder and the ambient fluid:

$$
-k
\left.
\frac{\partial T}{\partial r}
\right|_{r=R}
=
h\left(T-T_\infty\right).
$$

### Axial symmetry plane

Using symmetry about the mid-plane,

$$
\left.
\frac{\partial T}{\partial z}
\right|_{z=0}=0.
$$

### End surface

At \(z=L_z\),

$$
-k
\left.
\frac{\partial T}{\partial z}
\right|_{z=L_z}
=
h\left(T-T_\infty\right).
$$

## Analytical Benchmark

Define the dimensionless temperature as

$$
\theta = T-T_\infty,
$$

with

$$
\theta_i=T_i-T_\infty.
$$

The analytical solution for the finite cylinder can be obtained using the product of the infinite-cylinder and plane-wall solutions:

$$
\frac{\theta(r,z,t)}{\theta_i}
=
\left[
\frac{\theta_{\text{cyl}}(r,t)}{\theta_i}
\right]
\left[
\frac{\theta_{\text{slab}}(z,t)}{\theta_i}
\right].
$$

The cylindrical component is obtained from the analytical series solution developed for the radial cylinder problem, while the axial component is obtained from the analytical series solution for the plane-wall problem.

The radial Fourier number is defined as

$$
Fo_R=\frac{\alpha t}{R^2}.
$$
